## Ejercicio 1 \- Variables y Operaciones

Crear variables para almacenar:

tu nombre
tu edad
tu ciudad
Luego imprimir un mensaje con el siguiente formato:
Hola, mi nombre es \_\_\_, tengo \_\_\_ años y vivo en \_\_\_\.

```
nombre = "Angel"
edad = 43
ciudad = "Gerli, Lanús"

print(f"Hola, mi nombre es {nombre}, tengo {edad} años y vivo en {ciudad}.")
```

 A continuación, crear dos variables numéricas a y b, asignarles valores y calcular:

suma
resta
multiplicación
división
Mostrar los resultados utilizando print\(\)\.

```
num1 = 3750
num2 = 1250

suma = num1 + num2
resta = num1 - num2
multiplicacion = num1 * num2
division = num1 // num2

print(f"La suma es: {suma}")
print(f"La resta es: {resta}")
print(f"La multiplicación es: {multiplicacion}")
print(f"La división es: {division}")
```

## Ejercicio 2 \- Condicionales

Crear un programa que:

Pida al usuario que ingrese su edad usando input\(\)
Determine si la persona es mayor o menor de edad
Mostrar uno de los siguientes mensajes:

Eres mayor de edad
o
Eres menor de edad

```
edad = int(input("Por favor, ingresa tu edad: "))

if edad >= 18:
    print("Eres mayor de edad")
else:
    print("Eres menor de edad")
```

## Ejercicio 3 \- Listas y Bucles

Crear una lista con 5 números\. Utilizando un bucle for, imprimir cada número de la lista\. Mostrar ambos resultados por pantalla\.

```
lista = [25, 78, 101, 65, 99]

for i in lista:
    print(i)
```

Luego:

Calcular la suma total de los números

```
sum_total = sum(lista)
print("La suma total de los números es:", sum_total)
```


Calcular el promedio

```
prom_lista = sum(lista) / len(lista)
print(prom_lista)
```
