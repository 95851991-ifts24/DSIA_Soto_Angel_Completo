## Ejercicio 2 \- Dataset simple de Pandas

Crear un data set con la siguiente información:

Estudiante	Nota
Ana	                   8
Juan	           6
Carla	           9
Pedro	           5
Lucia	           7

```
df = pd.DataFrame(
    {
    'Estudiante': ['Ana','Juan','Carla','Pedro','Lucia'],
    'Nota': [8,6,9,5,7]
    })

# Mostrar el DataFrame por pantalla.
print(df)

# Calcular el promedio de las notas
df['Nota'].mean()
print(f'El promedio de notas es: {df['Nota'].mean()}')

# Mostrar la nota máxima
df['Nota'].max()
print(f'La nota máxima es: {df['Nota'].max()}')

# Mostrar la nota mínima
df['Nota'].min()
print(f'La nota mínima es: {df['Nota'].min()}')
```
